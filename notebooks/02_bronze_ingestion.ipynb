# 02 — Bronze Ingestion

**Week:** 4

## Goal

Read approved FitPulse batch source files from Unity Catalog Volume and load them into persistent Bronze Delta tables while preserving original source business values and ingestion metadata.

```python
USE CATALOG `fitpulse-wellness`;
USE SCHEMA default;
```

```python
## Source Inventory

| Source File | Format | Bronze Table |
|-------------|--------|-------------------------|
| activity_types.csv | CSV | bronze_activity_types |
| devices.csv | CSV | bronze_devices |
| goals.csv | CSV | bronze_goals |
| users.json | JSON | bronze_users |
| workouts.parquet | Parquet | bronze_workouts |
```

```python
LIST '/Volumes/fitpulse-wellness/default/fitpulsehub';
```

```python
CREATE OR REPLACE TABLE bronze_activity_types
USING DELTA
AS
SELECT *,
       _metadata.file_name AS source_file,
       current_timestamp() AS ingestion_timestamp,
       'RUN001' AS ingestion_run_id
FROM csv.`/Volumes/fitpulse-wellness/default/fitpulsehub/activity_types.csv`;
```

```python
SELECT *
FROM bronze_activity_types
LIMIT 10;
```

```python
SELECT COUNT(*) AS source_count
FROM csv.`/Volumes/fitpulse-wellness/default/fitpulsehub/activity_types.csv`;

SELECT COUNT(*) AS bronze_count
FROM bronze_activity_types;
```

```python
CREATE OR REPLACE TABLE bronze_devices
USING DELTA
AS
SELECT *,
       _metadata.file_name AS source_file,
       current_timestamp() AS ingestion_timestamp,
       'RUN001' AS ingestion_run_id
FROM csv.`/Volumes/fitpulse-wellness/default/fitpulsehub/devices.csv`;
```

```python
SELECT * FROM bronze_devices LIMIT 10;

SELECT COUNT(*) FROM csv.`/Volumes/fitpulse-wellness/default/fitpulsehub/devices.csv`;

SELECT COUNT(*) FROM bronze_devices;
```

```python
CREATE OR REPLACE TABLE bronze_goals
USING DELTA
AS
SELECT *,
       _metadata.file_name AS source_file,
       current_timestamp() AS ingestion_timestamp,
       'RUN001' AS ingestion_run_id
FROM csv.`/Volumes/fitpulse-wellness/default/fitpulsehub/goals.csv`;
```

```python
SELECT * FROM bronze_goals LIMIT 10;

SELECT COUNT(*) FROM csv.`/Volumes/fitpulse-wellness/default/fitpulsehub/goals.csv`;

SELECT COUNT(*) FROM bronze_goals;
```

```python
CREATE OR REPLACE TABLE bronze_users
USING DELTA
AS
SELECT *,
       _metadata.file_name AS source_file,
       current_timestamp() AS ingestion_timestamp,
       'RUN001' AS ingestion_run_id
FROM json.`/Volumes/fitpulse-wellness/default/fitpulsehub/users.json`;
```

```python
SELECT * FROM bronze_users LIMIT 10;

SELECT COUNT(*) FROM json.`/Volumes/fitpulse-wellness/default/fitpulsehub/users.json`;

SELECT COUNT(*) FROM bronze_users;
```

```python
CREATE OR REPLACE TABLE bronze_workouts
USING DELTA
AS
SELECT *,
       current_timestamp() AS ingestion_timestamp,
       'RUN001' AS ingestion_run_id
FROM parquet.`/Volumes/fitpulse-wellness/default/fitpulsehub/workouts.parquet`;
```

```python
SELECT *
FROM bronze_workouts
LIMIT 10;

SELECT COUNT(*) AS bronze_count
FROM bronze_workouts;
```

```python
SHOW TABLES IN `fitpulse-wellness`.default;
```

```python
DESCRIBE DETAIL bronze_activity_types;

DESCRIBE DETAIL bronze_devices;

DESCRIBE DETAIL bronze_goals;

DESCRIBE DETAIL bronze_users;

DESCRIBE DETAIL bronze_workouts;
```
